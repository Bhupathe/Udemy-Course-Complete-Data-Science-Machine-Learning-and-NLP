# Magic Methods

Magic methods in python, also known as dunder methods (double underscore methods), are special methods that start and end with double upderscores. These methods enable you to define the behaviour of objects for built-in operations, such as arithmetic operations, comparisons and more.

## Magic Methods
Magic methods are predefined methods in Python that you can override to change the behaviour of your objects. Some common magic methods include:

```python
'''
__init__ : Initializes a new instance of a class
__str__ : Returns a string representation of the object
__repr__ : Returns an official or unambiguous string representation of the object
__len__ : Returns the length of the object
__getitem__ : Gets an item from a container or object using indexing
__setitem__ : Sets an item in a container or object using indexing
__delitem__ : Deletes an item from a container or object using indexing
'''
```

```python
## __repr__ is used to define a string representation of an object that is unambiguous and can be used to 
## recreate the object. It is often used for debugging purposes. The __str__ method, on the other hand, is used to 
## define a string representation of an object that is more user-friendly and readable.

class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    
    def __repr__(self):
        return f"Person(name='{self.name}', age={self.age}, gender='{self.gender}')"


# Creating an instance of the Person class
person = Person("Alice", 30, "Female")
print(repr(person))  # Output: Person(name='Alice', age=30, gender
print(person)  # Output: Person(name='Alice', age=30, gender='Female')
```

```python
## __str__ is used to define a string representation of an object that is more user-friendly and readable. It is often used for displaying information to the end user. The __repr__ method, on the other hand, is used to define a string representation of an object that is unambiguous and can be used to recreate the object.

class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender

    def __str__(self):
        return f"{self.name}, {self.age} years old, {self.gender}"

# Creating an instance of the Person class
person = Person("Alice", 30, "Female")
print(repr(person))  
print(person)  # Output: Alice, 30 years old, Female
```
