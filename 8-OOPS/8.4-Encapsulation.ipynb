## Encapsulation
#### Public access specifier

```python
## Encapsulation with Getter and Setter Methods
### Keywords: Public, Private, Protected

class Person:
    def __init__(self, name, age):
        self.name = name  # Public attribute
        self.age = age   # Public attribute

def get_name(person):
    return person.name  # Accessing public attribute

person = Person("Alice", 30)
print(get_name(person))  # Output: Alice
```

#### Private Access Specifier

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name  # Private attribute
        self.__age = age   # Private attribute
        self.gender = gender  # Public attribute

## This is not a good practice to access private attributes directly. 
## Instead, we should use getter and setter methods to access and 
## modify private attributes.
def get_name(person):
    return person.__name  # Accessing private attribute

## This is not a good practice to access private attributes directly.
def get_age(person):
    return person._Person__age  # Accessing private attribute
```

```python
person = Person("Alice", 30, "Female")
dir(person)  # Output: ['_Person__name', '_Person__age', 'gender
```

```python
get_name(person)  # Output: Alice
```

```python
## This should not be accessed directly, but for demonstration purposes, 
## we can access it using the name mangling technique.
get_age(person)  # Output: 30
```

#### Protected Access Specifier

```python
class Person:
    def __init__(self, name, age, gender):
        self._name = name  # Protected attribute
        self._age = age   # Protected attribute
        self.gender = gender  # Public attribute

class Employee(Person):
    def __init__(self, name, age, gender, employee_id):
        super().__init__(name, age, gender)
        self.employee_id = employee_id  # Public attribute

# Creating an instance of Employee
emp = Employee("Bob", 28, "Male", "E12345")

# Accessing protected attributes from the subclass
print(emp._name)  # Output: Bob
print(emp._age)   # Output: 28
```

#### Encapsulation with Getter and Setter methods

```python
class Person:
    def __init__(self, name, age, gender):
        self.__name = name  # Private attribute
        self.__age = age   # Private attribute
        self.gender = gender  # Public attribute

    def get_name(self):
        return self.__name  # Accessing private attribute
    
    def set_name(self, name):
        self.__name = name  # Modifying private attribute

    def get_age(self):
        return self.__age  # Accessing private attribute
    
    def set_age(self, age):
        if age >= 0:
            self.__age = age  # Modifying private attribute
        else:
            raise ValueError("Age cannot be negative")
    
# creating instance of Person
person = Person("Alice", 30, "Female")

# Access and modify private attributes using getter and setter methods
print(person.get_name())  # Output: Alice
print(person.get_age())   # Output: 30

person.set_age(35)
print(person.get_age())   # Output: 35

try:
    person.set_age(-5)  # This will raise ValueError
except ValueError as e:
    print(e)  # Output: Age cannot be negative
```
