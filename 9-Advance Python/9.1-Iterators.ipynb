# Iterators in Python

Iterators are advanced Python concepts that allow for efficient looping and memore management. Iterators probide a way to access elements of a collection sequentially without exposing the underlying structure.

```python
lst = [1, 2, 3, 4, 5, 6]
for i in lst:
    print(i)
```

```python
type(lst)
```

```python
lst
```

```python
# creating an iterator from a list
iterator = iter(lst)
print(type(iterator))
```

```python
# Internally, I cannot see the elements.
# It uses lazy loading technique to load the elements one by one.
iterator
```

```python
## Iterate through all the elements using next() function

next(iterator) # run this untill it throws StopIteration error
```

```python
## Usage for String datatype

str_val = "Python"

iterator_str = iter(str_val)
```

```python
try:
    print(next(iterator_str)) # run this untill it throws StopIteration error
except StopIteration:
    print("End of string reached")
```
