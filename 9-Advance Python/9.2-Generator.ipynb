# Generators

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily, which means they generate values on the fly and do not store them in memory.

```python
def square(n):
    for i in range(n):
        return i**2
```

```python
square(3)
```

```python
def square(n):
    for i in range(n):
        yield i**2  # creates a local variable and store the value of i**2 in it. 
        # It will return the value of i**2 and pause the execution of the function. 
        # When next() is called, it will resume from where it left off and continue 
        # executing the function until it reaches the end or encounters another yield 
        # statement.
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
# or instead of using for loop, we can use next() function to get the next value 
# from the generator.

generator = square(3)
generator
```

```python
next(generator) # run this untill it throws StopIteration error
```

```python
def my_gen():
    yield 1
    yield 2
    yield 3
    yield 4
```

```python
generator = my_gen()
generator
```

```python
next(generator)
```

```python
for i in generator:
    print(i)
```

#### Practical Examples: Reading Large Files

Generators are particularly useful for reading large files because they allow you to process one line at a time without loading the entire file into memory.

```python
### Reading Large Files with Generators

def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line  # yield each line of the file one at a time
```

```python
file_path = 'large_file.txt'  # replace with the path to your large file

for line in read_large_file(file_path):
    print(line.strip())  # process each line (e.g., print it)
```
