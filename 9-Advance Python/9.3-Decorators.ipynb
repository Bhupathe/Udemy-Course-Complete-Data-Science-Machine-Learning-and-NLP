# Decorators

Decorators are a powerful and flexible feature in Python that allows you to modify the behaviour of a function or class method. They are commonly used to add functionality to functions or methods without modifying their actual code. This lesson covers the basics of decorators, including how to create and use them.

```python
### function copy
### closures
### decorators
```

```python
## Function copy
def welcome():
    return "Welcome to the world of Python!"
```

```python
wel = welcome
print(welcome)
print(wel)
print(wel())
```

```python
## Trying to delete the original function
del welcome
```

```python
## Am I still able to call the function using the copy?
print(wel())
```

```python
## Closures: Meaning function inside another function

def main_welcome():
    msg = "Welcome to the world of Python!"
    def inner_welcome():
        print(msg)
        
        print("Please enjoy your stay!")
    
    return inner_welcome()
    
```

```python
main_welcome()
```

```python
## Passing function as an argument

def main_welcome(func):
    def inner_welcome():
        func("Welcome to Advance Python course.")
        print("Please enjoy your stay!")
    
    return inner_welcome()
```

```python
main_welcome(print)
```

#### Decorators

```python
def main_welcome(func):
    def inner_welcome():
        func()
        print("Please enjoy your stay!")
    
    return inner_welcome()
```

```python
def course_introduction():
    print("This is learning course")

course_introduction()
```

```python
main_welcome(course_introduction)
```

Our main objective is to get above input without calling "main_welcome()" method

```python
@main_welcome
def course_introduction():
    print("This is learning course")
```

#### Example 2

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called")
        func()
        print("Something is happening after the function is called.")
    
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello!")
```

```python
say_hello()
```

#### Decorators with arguments

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("hello")
```

```python
say_hello()
```
