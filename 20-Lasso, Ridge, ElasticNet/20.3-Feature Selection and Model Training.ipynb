## Feature Selection

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'D:\Udemy Courses\complete_data_science_machine_learning_dl_nlp_bootcamp\20-Lasso, Ridge, ElasticNet\datasets\self\Algerian_forest_fires_dataset_UPDATE_modified.csv')
df.head()
```

```python
df.columns
```

```python
df.drop(['day', 'month', 'year'], axis=True, inplace=True)
```

```python
df['Classes'] = np.where(df['Classes'].str.contains('not fire'), 0, 1)
df.head()
```

```python
df['Classes'].value_counts()
```

```python
## Independent and Dependent Features
X = df.drop('FWI', axis = 1)
y = df['FWI']
```

```python
## Train, Test, Split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

negatively correlated can be important feature but highly positive correlated, you should remove it

```python
## Feature Selection based on Correlation
X_train.corr()
```

Multicollinearity means the independent features should not be that much correlated based on some threshold values

```python
## Check for Multicollinearity
plt.figure(figsize=(12, 10))
corr = X_train.corr()
sns.heatmap(corr, annot=True)
```

```python
def correlation(dataset, threshold):
    col_corr = set()
    corr_matrix = dataset.corr()
    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i, j]) > threshold:
                colname = corr_matrix.columns[i]
                col_corr.add(colname)
    
    return col_corr
```

```python
## Threshold -- Set up by Domain Expertise
corr_features = correlation(X_train, 0.85)
corr_features
```

```python
## Drop Features when correlation is more than 0.85
X_train.drop(corr_features, axis=1, inplace=True)
X_test.drop(corr_features, axis=1, inplace=True)
```

### Feature Scaling or Standardization

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Box Plots to Understand Effect of Standard Scalar

```python
plt.subplots(figsize=(15, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data=X_train)
plt.title('X_train Before Scaling')

plt.subplot(1, 2, 2)
sns.boxplot(data=X_train_scaled)
plt.title('X_train After Scaling')
```

### Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, r2_score

linear_reg = LinearRegression()
linear_reg.fit(X_train_scaled, y_train)

y_pred = linear_reg.predict(X_test_scaled)

mae = mean_absolute_error(y_test, y_pred)
score = r2_score(y_test, y_pred)
rev_score = r2_score(y_pred, y_test)

print(f"Mean Absolute Error: {mae}")
print(f"R2 Score: {score}")
```

```python
plt.scatter(y_test, y_pred)
plt.title("Linear Regression Model")
plt.xlabel("Y Test data")
plt.ylabel("Y Train Data")
```

## Lasso Regression
