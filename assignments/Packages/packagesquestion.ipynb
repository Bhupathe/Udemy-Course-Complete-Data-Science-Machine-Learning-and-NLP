### Assignment 9: Working with the `datetime` Module

Use the `datetime` module to print the current date, calculate the date 100 days from today, and determine the day of the week for a given date.

```python
#Your Code Here
from datetime import datetime, timedelta, date

today_date = datetime.now().date()
date_From_100 = today_date + timedelta(days=100)
given_date =date(2022, 1, 1)

print(f"Today's Date: {today_date}")
print(f"100 days from today: {date_From_100}")
print(f"Day of the week for 2022-01-01: {given_date.strftime('%A')}")
```

### Assignment 10: Working with the `random` Module

Use the `random` module to generate a list of 5 random numbers between 1 and 50 and shuffle the elements of a list.

```python
#Your Code Here
import random

lst = [random.randint(1, 50) for _ in range(5)]
print(f"Before Shuffle: {lst}")

random.shuffle(lst)
print(f"After Shuffle: {lst}")
```
